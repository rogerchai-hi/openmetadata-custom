```python
from scipy.stats import linregress
from sklearn.metrics import mean_absolute_error
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_parquet(r"data\traffic_data.parquet")
```

# 漂移前訓練模型

```python
df_train = df[df["year"] == 113]
X = df_train["accident_count"]
Y = df_train["injury_count"]

slope, intercept, r, p, std_err = linregress(X, Y)

def predict(x):
    return [slope * xi + intercept for xi in x]
```

```python
mae = mean_absolute_error(Y, predict(X))
print(f"MAE: {mae:.2f}")
```

# 漂移後新族群樣態

```python
df_predict = df[df["year"] == 114][["accident_count", "injury_count"]]
X = df_predict["accident_count"]
Y = df_predict["injury_count"]

mae = mean_absolute_error(Y, predict(X))
print(f"MAE: {mae:.2f}")
```

```python
df_plot = df.copy()
df_plot["year"] = df_plot["year"].astype(str)
fig = px.scatter(df_plot,
                 x = "accident_count",
                 y = "injury_count",
                 color = "year"
)
fig.update_layout(
    width=600,
    height=500,

)
fig.show()
```
